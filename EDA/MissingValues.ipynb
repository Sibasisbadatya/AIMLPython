Strategy MultiVariate Imputation
Here we cant replace the missing valus with the independent mean or statistice of that single column only rather we have to consider
relatioship between other columns depends on also.

We can use IterativeImputer or KNN Imputer



👉 SimpleImputer fills missing values using one simple rule
👉 IterativeImputer fills missing values by learning patterns from other features


| Age | Salary |
| --- | ------ |
| 25  | 30000  |
| 30  | 40000  |
| 35  | NaN    |

Now salary is missing.

SimpleImputer
might fill with median salary:
35000

IterativeImputer

looks at pattern:

age 25 → salary 30000
age 30 → salary 40000

Then predicts:
age 35 → salary ≈ 50000
This is smarter.



IterativeImputer takes one target column where it should replace values and othres as supportive to calculate so we first handle the supoort column misisng values with mean,median or mode or any other


```python
df = pd.DataFrame({
    'MedInc': [8.3, 8.0, np.nan, 5.5, 3.8, np.nan, 7.0],   # Median Income
    'HouseAge': [41, 21, 52, np.nan, 52, 15, 30],          # Age of the house
    'MedHouseVal': [4.5, 3.5, 3.5, 3.4, np.nan, 2.8, 3.0] # Median House Value
})

df
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

imp = IterativeImputer()

# Create the imputer object.

# By default it:
# uses a regression model internally
# predicts missing values using other columns
# repeats multiple times to refine predictions

df_imputed = pd.DataFrame(imp.fit_transform(df), columns=df.columns)
# This is the main step.

# 1. imp.fit_transform(df)
# looks at all columns
# finds missing values
# predicts them using other columns
# returns a filled NumPy array

# 2. pd.DataFrame(..., columns=df.columns)
# fit_transform() returns only a NumPy array, so column names are lost.
# This converts it back into a DataFrame and restores original column names.

df_imputed
```

KNNImputer in scikit-learn fills missing values by looking at similar rows.

KNN = K Nearest Neighbors

So the idea is:

👉 find rows that look similar
👉 use them to fill the missing value

Simple intuition

Suppose one student’s math marks are missing.

You look for students with similar:

science marks
english marks
attendance

Then estimate the missing math mark using those similar students.

That is exactly how KNNImputer works.


How it works
--------------------------------

For a missing value:

Step 1

Find rows similar to this row

Step 2

Pick nearest K rows
(K = number of neighbors)

Step 3

Use their values to fill missing value

Usually by averaging them.

----------------------------------------------------------------
How Both Itertive Imputer Works
----------------------------------------------------------------

| Age | Experience | Salary |
| --- | ---------- | ------ |
| 25  | 2          | 30000  |
| 30  | 5          | 50000  |
| 35  | 7          | NaN    |

Salary is missing.

Instead of filling with average salary, IterativeImputer thinks:

👉 Salary probably depends on:

Age
Experience

So it learns pattern:
Salary = f(Age, Experience)

Step-by-step working
Step 1: Initial fill (rough guess)

First, it quickly fills all missing values with simple estimates
(like mean).

So missing values get temporary placeholders.

| Age | Experience | Salary    |                   |
| --- | ---------- | --------- | ----------------- |
| 25  | 2          | 30000     |                   |
| 30  | 5          | 50000     |                   |
| 35  | 7          | **40000** | ← temporary guess |

Step 2: Pick one column with missing values

Now it selects one column (say Salary) and treats it as target.

It says:

👉 let me predict Salary using other columns

So:

target = Salary
inputs = Age, Experience


Step 3: Train a model

It trains a small regression model using rows where Salary is known.

From known rows:

| Age | Experience | Salary |
| --- | ---------- | ------ |
| 25  | 2          | 30000  |
| 30  | 5          | 50000  |


It learns relation:

👉 higher age + more experience → higher salary

Step 4: Predict missing value

Now it predicts missing Salary for:

Age	Experience
35	7

Now temporary 40000 gets replaced with smarter 65000.

Step 5: Move to next column

Then it repeats same process for another column with missing values.

Example:

predict missing Age using Experience, Salary
predict missing Experience using Age, Salary

So every column gets improved one by one.

-----------------------------------------------------------------
That is exactly how KNNImputer works.
--------------------------------------------------------------------


| Age | Experience | Salary |
| --- | ---------- | ------ |
| 25  | 2          | 30000  |
| 30  | 5          | 50000  |
| 35  | 6          | NaN    |

Salary is missing in row 3.

We want to fill it.
Step-by-step how KNNImputer works

Step 1: Find the row with missing value
| Age | Experience | Salary |
| --- | ---------- | ------ |
| 35  | 6          | NaN    |


Step 2: Compare this row with other rows

Now compare this row with rows where Salary is known.

| Age | Experience |
| --- | ---------- |
| 25  | 2          |

Difference from missing row (35, 6) is large.

So row 1 is not very similar.

Compare with row 2

| Age | Experience |
| --- | ---------- |
| 30  | 5          |

This is much closer to (35, 6).

So row 2 is more similar.

Step 3: Find nearest neighbors

Now choose the nearest rows (most similar rows).
if n_neighbors = 2

take the 2 closest rows.

These are the K nearest neighbors.

Now look at their Salary values:

row 1 → 30000
row 2 → 50000

Use them to estimate missing Salary.

Usually by averaging.
(30000 + 50000) / 2 = 40000

Missing Salary becomes 40000.






| Aspect                           | IterativeImputer                            | KNNImputer                                     |
| -------------------------------- | ------------------------------------------- | ---------------------------------------------- |
| **How it fills missing values**  | Predicts missing values using other columns | Fills using values from similar rows           |
| **Core idea**                    | Learns feature relationships                | Finds nearest neighbors                        |
| **Method type**                  | Model-based (regression)                    | Distance-based                                 |
| **Looks at**                     | Column relationships                        | Row similarity                                 |
| **Best when**                    | Features are correlated                     | Similar rows exist                             |
| **Example thinking**             | “Predict salary using age + experience”     | “Find similar people and use their salary”     |
| **Captures feature dependency?** | Yes                                         | Partially                                      |
| **Uses global pattern?**         | Yes (whole dataset)                         | No (local neighbors only)                      |
| **Uses local similarity?**       | No                                          | Yes                                            |
| **Speed**                        | Moderate                                    | Slower on large data                           |
| **Scaling needed?**              | Optional                                    | Very important                                 |
| **Why scaling matters**          | Model can adjust weights                    | Distance gets dominated by large-scale columns |
| **Output quality**               | Better when columns are related             | Better when similar rows exist                 |
| **Good for**                     | Structured/tabular correlated data          | Clustered / similarity-based data              |
| **Main strength**                | Smarter feature-based prediction            | Natural neighbor-based filling                 |
| **Main weakness**                | Slower, more complex                        | Slow + sensitive to scaling                    |
