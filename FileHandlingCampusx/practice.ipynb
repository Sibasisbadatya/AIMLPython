## Using context manager(with)

```python
#with
with open("sample3.txt", "w") as f:
    f.write("Hello World")
# with automatically closes the file after use so we dont need to explicitly close the file
```

```python
with open("sample3.txt", "r") as f:
    print(f.read(10))
    print(f.read(10))
```

```python
# Loading big file in memory using readlines
bigL = ['hello world' for _ in range(1000000)]
with open("bigfile.txt", "w") as f:
    f.writelines(bigL)
```

```python

with open("bigfile.txt", "r") as f:
    chunk_size = 100
    while True:
        chunk = f.read(chunk_size)
        if not chunk:
            break

        print(chunk, end="***")
    
```

Seek and Tell Funciton

```python
with open("sample3.txt", "r") as f:
    print(f.read(10))
    print(f.tell())
    # tell gives the current position of the file pointer in the file
```

```python
with open("sample3.txt", "r") as f:
    print(f.read(10))
    print(f.tell())
    f.seek(0)
    # moves the current read cursor to start position (for this code) of the file
    # but with the help of seek we can move anywhere in the file
    print(f.read(10))
```

tell and seek in write mode

```python
f = open("sample4.txt", "w")
f.write("Hello, World!")
f.seek(0)
f.write("Python")
f.close()
```

serialise and deserialise with custom class and object

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

person = Person("Sibasis",30)
person
with open("sample3.json","w") as f:
    json.dump(person,f)
# above will give an error as json does not know how to serialize the custom objects like class objects
    
```

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
def person_serializer(obj):
    if isinstance(obj, Person):
        return {"name": obj.name, "age": obj.age}
    raise TypeError("Type not serializable")

def show_person_details(person):
    if isinstance(person, Person):
        return "{} is {} years old".format(person.name, person.age)
    raise TypeError("Type not serializable")

person = Person("Sibasis",30)
person
with open("sample3.json","w") as f:
    json.dump(person,f,default=person_serializer)
# above will give an error as json does not know how to serialize the custom objects like class objects
    
```

```python
# deserialising

with open("sample3.json","r") as f:
    data = json.load(f)
    print(data)
    print(type(data))
```

Pickling and UnPickling

```python
class Person:

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('Hi my name is',self.name,'and I am ',self.age,'years old')
    
    
```

```python
p = Person('nitish',33)
```

```python
# pickle dump
import pickle
with open('person.pkl','wb') as f:
  pickle.dump(p,f)
```

```python
# pickle load
import pickle
with open('person.pkl','rb') as f:
  p = pickle.load(f)

p.display_info()
```
